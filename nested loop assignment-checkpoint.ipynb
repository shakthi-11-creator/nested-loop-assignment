## hallow square

```python
def hollow_square(n):
    for i in range(n):
        for j in range(n):
            if i == 0 or i == n-1 or j == 0 or j == n-1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_square(5)
```

## Hallow Rectangle

```python
def hollow_rectangle(rows, cols):
    for i in range(rows):
        for j in range(cols):
            if i == 0 or i == rows-1 or j == 0 or j == cols-1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_rectangle(4, 7)
```

## Hallow Right-angled triangle

```python
def hollow_right_triangle(n):
    for i in range(1, n+1):
        for j in range(1, i+1):
            if j == 1 or j == i or i == n:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_right_triangle(5)
```

## Hallow Inverted Triangle

```python
def hollow_inverted_triangle(n):
    for i in range(n, 0, -1):
        for j in range(1, i+1):
            if j == 1 or j == i or i == n:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_inverted_triangle(5)
```

## Hallow Pyramid

```python
def hollow_pyramid(n):
    for i in range(1, n+1):
        for j in range(n-i):
            print(" ", end=" ")
        for j in range(1, 2*i):
            if j == 1 or j == 2*i-1 or i == n:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_pyramid(5)
```

## Hallow inverted pyramid

```python
def hollow_inverted_pyramid(n):
    for i in range(n, 0, -1):
        for j in range(n-i):
            print(" ", end=" ")
        for j in range(1, 2*i):
            if j == 1 or j == 2*i-1 or i == n:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_inverted_pyramid(5)
```

## Hallow Diamond

```python
def hollow_diamond(n):
    # Upper half
    for i in range(1, n+1):
        for j in range(n-i):
            print(" ", end=" ")
        for j in range(1, 2*i):
            if j == 1 or j == 2*i-1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()
    # Lower half
    for i in range(n-1, 0, -1):
        for j in range(n-i):
            print(" ", end=" ")
        for j in range(1, 2*i):
            if j == 1 or j == 2*i-1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_diamond(5)
```

## hallow hourglass

```python
def hollow_hourglass(n):
    # Upper inverted pyramid
    for i in range(n, 0, -1):
        for j in range(n-i):
            print(" ", end=" ")
        for j in range(1, 2*i):
            if j == 1 or j == 2*i-1 or i == n:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()
    # Lower pyramid
    for i in range(2, n+1):
        for j in range(n-i):
            print(" ", end=" ")
        for j in range(1, 2*i):
            if j == 1 or j == 2*i-1 or i == n:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_hourglass(5)
```

## hallow cross

```python
def hollow_cross(n):
    for i in range(n):
        for j in range(n):
            if i == n//2 or j == n//2:
                print("+", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_cross(7)
```

## Hallow X

```python
def hollow_x(n):
    for i in range(n):
        for j in range(n):
            if i == j or i + j == n-1:
                print("*", end=" ")
            else:
                print(" ", end=" ")
        print()

hollow_x(7)
```
